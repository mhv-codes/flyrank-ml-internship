## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset from your GitHub repository
url = "https://raw.githubusercontent.com/mhv-codes/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# 1. Dataset size
print("Dataset rows:", len(df))
print("Dataset columns:", len(df.columns))

# 2. Number of distinct clients
print("Distinct anonymized clients:", df["client_id"].nunique())

# 3. Content trend distribution
print("\nContent trend distribution:")
print(df["trend_direction"].value_counts(dropna=False))

# 4. Compare content age with trend direction
print("\nTrend distribution by content age:")
age_groups = pd.qcut(df["content_age_days"], q=4, duplicates="drop")
print(pd.crosstab(age_groups, df["trend_direction"], normalize="index").round(3))
```
