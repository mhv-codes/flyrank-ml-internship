## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: Precision@K

I will use Precision@K as the primary evaluation metric for the ranking task. It measures the proportion of relevant pages among the top K pages recommended for review.

For example, Precision@50 would measure how many of the top 50 recommended pages are confirmed as useful review opportunities.

Since the starter dataset does not contain verified editor-confirmed review labels, actual Precision@50 cannot be calculated yet. It will require human validation of the recommendations.

A higher Precision@K would indicate that editors are more likely to find useful opportunities near the top of the ranked list.
